```python
from pathlib import Path

DATASET_ROOT = Path("/content/data/AID")

print("Dataset exists:", DATASET_ROOT.exists())

classes = sorted(
    p.name for p in DATASET_ROOT.iterdir()
    if p.is_dir()
)

print("Number of classes:", len(classes))
print("Classes:", classes)
```

```python
import random
import time
import json
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split, Subset
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, confusion_matrix,
    ConfusionMatrixDisplay
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
print("PyTorch:", torch.__version__)
```

```python
IMAGE_SIZE = 128
BATCH_SIZE = 32

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize((0.5,)*3, (0.5,)*3)
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,)*3, (0.5,)*3)
])

base_dataset = datasets.ImageFolder(
    str(DATASET_ROOT), transform=eval_transform
)

class_names = base_dataset.classes
num_classes = len(class_names)

print("Total images:", len(base_dataset))
print("Total classes:", num_classes)
print("Classes:", class_names)

assert num_classes > 1, "Check your dataset folder structure."
```

```python
n = len(base_dataset)
n_train = int(0.70 * n)
n_val = int(0.15 * n)
n_test = n - n_train - n_val

generator = torch.Generator().manual_seed(SEED)

train_split, val_dataset, test_dataset = random_split(
    base_dataset,
    [n_train, n_val, n_test],
    generator=generator
)

augmented_dataset = datasets.ImageFolder(
    str(DATASET_ROOT), transform=train_transform
)

train_dataset = Subset(
    augmented_dataset, train_split.indices
)

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE,
    shuffle=True, num_workers=2
)

val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE,
    shuffle=False, num_workers=2
)

test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE,
    shuffle=False, num_workers=2
)

print("Training:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Testing:", len(test_dataset))
```

```python
class AerialCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 192, 3, padding=1),
            nn.BatchNorm2d(192),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.35),
            nn.Linear(192, 128),
            nn.ReLU(),
            nn.Dropout(0.25),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        return self.classifier(self.features(x))

model = AerialCNN(num_classes).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print(model)
```

```python
EPOCHS = 10

history = {
    "train_loss": [],
    "val_loss": [],
    "train_acc": [],
    "val_acc": []
}

def run_epoch(loader, training):
    model.train() if training else model.eval()

    total_loss = 0.0
    correct = 0
    total = 0

    with torch.set_grad_enabled(training):
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            if training:
                optimizer.zero_grad()

            outputs = model(images)
            loss = criterion(outputs, labels)

            if training:
                loss.backward()
                optimizer.step()

            total_loss += loss.item() * labels.size(0)
            correct += (outputs.argmax(1) == labels).sum().item()
            total += labels.size(0)

    return total_loss / total, correct / total

start_time = time.time()

for epoch in range(EPOCHS):
    train_loss, train_acc = run_epoch(train_loader, True)
    val_loss, val_acc = run_epoch(val_loader, False)

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    print(
        f"Epoch {epoch+1}/{EPOCHS} | "
        f"Train Accuracy: {train_acc:.4f} | "
        f"Val Accuracy: {val_acc:.4f} | "
        f"Train Loss: {train_loss:.4f}"
    )

print("Training completed!")
print("Minutes:", round((time.time() - start_time) / 60, 2))
```

```python
epochs = range(1, EPOCHS + 1)

plt.figure(figsize=(8, 4))
plt.plot(epochs, history["train_acc"], label="Training")
plt.plot(epochs, history["val_acc"], label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("AID Training and Validation Accuracy")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(8, 4))
plt.plot(epochs, history["train_loss"], label="Training")
plt.plot(epochs, history["val_loss"], label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("AID Training and Validation Loss")
plt.legend()
plt.grid()
plt.show()
```

```python
model.eval()
y_true, y_pred = [], []

with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images.to(device))
        predictions = outputs.argmax(1).cpu()

        y_true.extend(labels.numpy())
        y_pred.extend(predictions.numpy())

test_acc = accuracy_score(y_true, y_pred)
precision = precision_score(
    y_true, y_pred, average="macro", zero_division=0
)
recall = recall_score(
    y_true, y_pred, average="macro", zero_division=0
)
f1 = f1_score(
    y_true, y_pred, average="macro", zero_division=0
)

print(f"Test Accuracy: {test_acc:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall: {recall:.4f}")
print(f"Macro F1 Score: {f1:.4f}")

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(num_classes)),
    target_names=class_names,
    zero_division=0
)

print(report)
```

```python
cm = confusion_matrix(
    y_true, y_pred,
    labels=list(range(num_classes))
)

fig, ax = plt.subplots(figsize=(14, 12))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot(
    ax=ax,
    xticks_rotation=90,
    colorbar=False
)

plt.title("AID Aerial Scene Classification")
plt.tight_layout()
plt.savefig("/content/aid_confusion_matrix.png", dpi=180)
plt.show()
```

```python
torch.save(
    model.state_dict(),
    "/content/aid_aerial_cnn.pth"
)

results = {
    "dataset": "AID Aerial Image Dataset",
    "classes": class_names,
    "epochs": EPOCHS,
    "test_accuracy": float(test_acc),
    "precision": float(precision),
    "recall": float(recall),
    "f1_score": float(f1),
    "history": history
}

with open("/content/aid_results.json", "w") as f:
    json.dump(results, f, indent=2)

with open("/content/aid_classification_report.txt", "w") as f:
    f.write(report)

print("Model and evaluation results saved!")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# 1. Training and Validation Accuracy
epochs = range(1, len(history["train_acc"]) + 1)

plt.figure(figsize=(9, 5))
plt.plot(epochs, history["train_acc"], marker="o", label="Training Accuracy")
plt.plot(epochs, history["val_acc"], marker="s", label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("AID Aerial Scene Classification - Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("/content/accuracy_graph.png", dpi=300)
plt.show()


# 2. Training and Validation Loss
plt.figure(figsize=(9, 5))
plt.plot(epochs, history["train_loss"], marker="o", label="Training Loss")
plt.plot(epochs, history["val_loss"], marker="s", label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("AID Aerial Scene Classification - Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("/content/loss_graph.png", dpi=300)
plt.show()


# 3. Confusion Matrix
model.eval()
y_true, y_pred = [], []

with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images.to(device))
        predictions = outputs.argmax(dim=1).cpu().numpy()

        y_true.extend(labels.numpy())
        y_pred.extend(predictions)

cm = confusion_matrix(
    y_true, y_pred,
    labels=list(range(len(class_names)))
)

fig, ax = plt.subplots(figsize=(14, 12))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)
disp.plot(ax=ax, xticks_rotation=90, colorbar=False)
plt.title("AID Aerial Scene Classification - Confusion Matrix")
plt.tight_layout()
plt.savefig("/content/confusion_matrix.png", dpi=300, bbox_inches="tight")
plt.show()


# 4. Final Classification Metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(
    y_true, y_pred, average="macro", zero_division=0
)
recall = recall_score(
    y_true, y_pred, average="macro", zero_division=0
)
f1 = f1_score(
    y_true, y_pred, average="macro", zero_division=0
)

metrics = [accuracy, precision, recall, f1]
labels = ["Accuracy", "Precision", "Recall", "F1 Score"]

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, metrics)

plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("Final AID Classification Performance")

for bar, value in zip(bars, metrics):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.02,
        f"{value:.3f}",
        ha="center"
    )

plt.tight_layout()
plt.savefig("/content/final_classification_metrics.png", dpi=300)
plt.show()

print("All four result images generated successfully!")
```
